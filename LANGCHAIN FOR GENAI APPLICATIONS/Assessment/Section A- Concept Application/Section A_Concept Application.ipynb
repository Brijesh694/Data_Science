### SCENARIO 1:- You are deploying a delivery time prediction model for a food delivery startup. The data science team trained the model in a Jupyter notebook using scikit-learn, and you need to ship it as part of a production Flask service that will handle several hundred requests per hour.

Question: Compare Pickle and Joblib as serialisation choices for this deployment context. Which would you use and why? What risks does each carry when the model is loaded back in an environment that differs from the one it was trained in?


| Feature              | Pickle                              | Joblib                                                 |
| -------------------- | ----------------------------------- | ------------------------------------------------------ |
| Purpose              | General Python object serialization | Optimized for NumPy arrays and machine learning models |
| Performance          | Slower for large models             | Faster loading/saving for large models                 |
| File Size            | Usually larger                      | Often smaller due to compression                       |
| Scikit-learn Support | Supported                           | Recommended by scikit-learn                            |
| Memory Efficiency    | Less efficient                      | Better handling of large arrays                        |


Which would I use?

For a production Flask service handling hundreds of requests per hour, I would use Joblib.

Reasons:

1. Most scikit-learn models contain large NumPy arrays (coefficients, trees, parameters).
2. Joblib is optimized for storing and loading these arrays efficiently.
3. Faster model loading reduces service startup time.
4. Optional compression decreases storage requirements.
5. It is the serialization method commonly recommended in scikit-learn documentation.

Risks of Pickle
1. Security Risk
    Loading a pickle file executes Python instructions.
    A malicious pickle file can run arbitrary code.
    Never load pickle files from untrusted sources.
2. Version Compatibility
    A model pickled using one version of scikit-learn may fail when loaded with another version.
     Internal class structures can change between releases.
3. Dependency Differences
    Different Python, NumPy, or scikit-learn versions may cause errors or incorrect behavior.

Risks of Joblib
1. Same Security Risk
Joblib internally relies on pickle.
Loading untrusted files can execute arbitrary code.
2. Version Compatibility
Models saved with one library version may not load correctly in another environment.
3. Dependency Mismatch
Different versions of NumPy, SciPy, or scikit-learn can lead to loading failures or prediction inconsistencies.

Handling Environment Differences

If the deployment environment differs from the training environment:

Use the same Python version.
Use the same scikit-learn, NumPy, and SciPy versions.
Store dependencies in a requirements.txt file.
Consider using Docker to package the exact training environment.
Record model metadata (library versions, training date, feature schema).

1. Architectural Differences
Variational Autoencoder (VAE)

A VAE consists of two main components:

Encoder: Compresses an image into a low-dimensional latent representation.
Decoder: Reconstructs an image from that latent representation.

Input Image
      ↓
   Encoder
      ↓
 Latent Space
      ↓
   Decoder
      ↓
 Generated Image

Stable Diffusion

Stable Diffusion is a latent diffusion model.

Text Prompt + Reference Image
            ↓
      Text Encoder
            ↓
     Latent Representation
            ↓
  Iterative Denoising Process
            ↓
      High-Quality Image

| Aspect                    | Variational Autoencoder (VAE)                                                                               | Stable Diffusion                                                                                                               |
| ------------------------- | ----------------------------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------------ |
| **Architecture**          | Encoder compresses images into a latent space; decoder reconstructs images from that latent representation. | Diffusion model gradually adds noise to images during training and learns to remove noise step-by-step to generate new images. |
| **Generation Process**    | Generates images directly from latent vectors.                                                              | Starts from random noise and iteratively denoises to create realistic images.                                                  |
| **Image Quality**         | Often produces blurry or less detailed outputs.                                                             | Produces highly detailed, photorealistic images with rich textures.                                                            |
| **Text Conditioning**     | Limited text-to-image capabilities without additional components.                                           | Built specifically for text-to-image generation using language-image embeddings.                                               |
| **Reference Image Usage** | Can reconstruct or slightly modify images but struggles with complex editing.                               | Supports image-to-image generation, style transfer, inpainting, and controlled editing using a reference image.                |
| **Training Complexity**   | Easier and faster to train.                                                                                 | More computationally expensive and requires larger datasets.                                                                   |
| **Diversity of Outputs**  | May generate less diverse samples due to latent space limitations.                                          | Generates highly diverse and creative outputs while maintaining realism.                                                       |
| **Commercial Readiness**  | Useful for representation learning, compression, and anomaly detection.                                     | Widely used in production-grade image generation systems.                                                                      |

### SCENARIO:-3 You are developing an AI assistant for a food delivery platform. The assistant must handle two distinct jobs: (1) answering open-ended questions such as 'What cuisines are trending near me this weekend?', and (2) classifying incoming customer reviews as Positive, Neutral, or Negative for the analytics dashboard.


Question: GPT-style decoder-only models and BERT-style encoder-only models are both available to your team. Which architecture is better suited to each job, and what fundamental property of each architecture makes it the right choice for that specific task?

| Task                            | Best Architecture   | Reason                                                 |
| ------------------------------- | ------------------- | ------------------------------------------------------ |
| Answering open-ended questions  | GPT (Decoder-Only)  | Designed for text generation                           |
| Review sentiment classification | BERT (Encoder-Only) | Designed for language understanding and classification |


1. Open-Ended Question Answering

Example:

"What cuisines are trending near me this weekend?"

Recommended Model: GPT (Decoder-Only)

GPT models are autoregressive, meaning they generate text one token at a time based on previously generated tokens.

Architecture:

Input Prompt
      ↓
 Decoder
      ↓
 Next Token Prediction
      ↓
 Generated Response
Fundamental Property

Causal (unidirectional) attention

Each token can attend only to previous tokens, allowing the model to predict and generate the next word in a sequence.

Example:

User: What cuisines are trending near me this weekend?

GPT:
Based on recent orders, Indian street food, Korean barbecue, and
artisan pizza are currently trending in your area...
Why GPT Fits
Produces natural language responses.
Handles conversational interactions.
Generates summaries, recommendations, and explanations.
Maintains context across multiple turns.

Therefore, GPT is ideal for generative AI assistant tasks.

2. Customer Review Classification

Example reviews:

"The food arrived hot and tasted amazing."

"Delivery was okay, nothing special."

"My order was late and completely cold."

Output:

Positive
Neutral
Negative
Recommended Model: BERT (Encoder-Only)

BERT processes the entire sentence simultaneously rather than generating text.

Architecture:

Review Text
      ↓
 Bidirectional Encoder
      ↓
 Sentence Representation
      ↓
 Classification Layer
      ↓
 Sentiment Label
Fundamental Property

Bidirectional attention

Each word can attend to both left and right context.

Example:

"The food was not good."

BERT understands that:

"not" modifies "good"
Meaning depends on both preceding and following words

This bidirectional understanding improves classification accuracy.

Why BERT Fits
Excellent language understanding.
Strong contextual representations.
Fast inference for classification tasks.
Requires only a small classification head on top of the encoder.

Therefore, BERT is ideal for sentiment analysis and review categorization.

Key Architectural Difference
1. GPT (Decoder-Only)
Past Tokens → Current Token → Future Tokens Hidden
>Unidirectional attention.
>Optimized for text generation.
>Learns next-token prediction.
2.  BERT (Encoder-Only)
Past Tokens ↔ Current Token ↔ Future Tokens
>Bidirectional attention.
>Optimized for understanding text.
>Learns contextual representations.

Few-shot prompting is the better choice.

It helps the model generalize to similar unseen complaints such as damaged packaging, missing cutlery, or incorrect items.

### SCENARIO :-5 You are designing a LangChain-based food ordering assistant. A simpler alternative is a single LLM call that takes the user's message and returns a response directly. Your team is debating whether to build a full agent with tools or use a straightforward LLMChain.


Question: Describe what distinguishes a LangChain agent from a simple LLM chain. Given that your food ordering assistant must check live restaurant availability, call an API to calculate estimated delivery time, and validate discount codes, justify which approach is more appropriate — and identify two risks the agent approach introduces.

| Feature        | Simple LLM Chain        | LangChain Agent                        |
| -------------- | ----------------------- | -------------------------------------- |
| Execution Flow | Fixed sequence of steps | Dynamic decision-making                |
| Tool Usage     | Predefined by developer | Chooses tools at runtime               |
| Flexibility    | Limited                 | High                                   |
| Reasoning      | Single prompt → output  | Reason → act → observe → repeat        |
| Best For       | Predictable workflows   | Complex tasks requiring multiple tools |


1. Simple LLM Chain

A simple chain follows a predefined workflow.

User Query
    ↓
Prompt Template
    ↓
LLM
    ↓
Response
If a customer asks:

"Can I order a pizza from Restaurant A with code SAVE20?"

A chain can generate text but cannot independently decide to:

Check restaurant availability
Call a delivery-time API
Validate the discount code

unless these actions are explicitly hardcoded into the workflow.

Advantage
Simple to build
Predictable behavior
Faster execution
Limitation
Cannot dynamically decide which external tools to use.

2. LangChain Agent

An agent has access to tools and can decide which ones to invoke.

Example workflow:

User Request
      ↓
      Agent
      ↓
Reasoning
      ↓
Choose Tool
      ↓
Observe Result
      ↓
Choose Next Tool
      ↓
Final Answer

For example:

User:

"Can I get a burger from Burger House using code SAVE20, and how long will delivery take?"

The agent may:

Check restaurant availability.
Call the delivery-time API.
Validate SAVE20 discount code.
Combine results into a final response.

This is often called the Reason-Act-Observe loop.

| User Query                            | Tools Needed       |
| ------------------------------------- | ------------------ |
| "Is Restaurant A open?"               | Availability API   |
| "How long will delivery take?"        | Delivery API       |
| "Does SAVE20 work?"                   | Discount Validator |
| "Order from Restaurant A with SAVE20" | All three tools    |


### SCENARIO :-6 You are optimising a food delivery chatbot that must accurately answer questions about the platform's 500-page policy document and a menu catalogue that is updated daily. Your team is evaluating two approaches: fine-tuning the underlying LLM on this data, or implementing a Retrieval-Augmented Generation (RAG) pipeline.


Question: Compare fine-tuning and RAG for this specific scenario. Address update frequency,cost, hallucination risk, and data privacy. Which approach is more suitable for a daily-updated,document-grounded Q&A bot, and what limitations does your recommended approach still carry?

Fine-Tuning vs Retrieval-Augmented Generation (RAG)

For a food delivery chatbot that must answer questions about a 500-page policy document and a daily-updated menu catalogue, both fine-tuning and RAG can improve performance, but they solve different problems

| Aspect             | Fine-Tuning                                       | RAG                                       |
| ------------------ | ------------------------------------------------- | ----------------------------------------- |
| Purpose            | Teach the model new behaviors or domain knowledge | Retrieve relevant documents at query time |
| Knowledge Updates  | Requires retraining                               | Update knowledge base only                |
| Cost               | High training cost                                | Lower maintenance cost                    |
| Response Accuracy  | Depends on learned knowledge                      | Grounded in retrieved documents           |
| Hallucination Risk | Higher                                            | Lower                                     |
| Scalability        | Difficult for frequent updates                    | Well suited for changing data             |
| Data Freshness     | Limited to training date                          | Can use latest documents                  |
| Explainability     | Limited                                           | Can provide source references             |


For a food delivery chatbot that must answer questions from a 500-page policy document and a daily-updated menu catalogue, RAG is the preferred approach. Compared with fine-tuning, it handles frequent updates more efficiently, costs less to maintain, provides fresher information, improves privacy control, and significantly reduces hallucinations by grounding responses in retrieved documents. However, RAG still depends on high-quality retrieval, incurs additional latency, and cannot completely eliminate hallucinations, making careful document indexing and retrieval design essential for a reliable production system.
