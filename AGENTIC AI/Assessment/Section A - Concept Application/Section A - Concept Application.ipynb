# SCENARIO:-1 


You are building a chatbot for a food delivery platform that must recommend dishes to
users based on their dietary restrictions, past order history, and real-time menu
availability from partnered restaurants.


Question: Your team is debating whether to use prompt engineering, RAG, or fine-tuning to
power this chatbot. The menu data updates daily and the user base is diverse. Evaluate all three
approaches against this use-case and justify which combination — or single approach — you
would recommend, and why the others fall short for this specific problem.

For a food delivery chatbot that recommends dishes based on dietary restrictions, past order history, and real-time restaurant menu availability, each AI technique has different strengths.

1. Prompt Engineering

Prompt engineering improves the way an existing language model responds by providing clear instructions. It is simple to implement, requires no additional training, and is suitable for generating conversational responses or answering general customer queries.

Advantages:

Easy and inexpensive to implement.
No model retraining required.
Good for formatting responses and handling simple conversations.

Limitations:

Cannot access real-time menu availability on its own.
Does not remember user history unless it is provided in the prompt.
Less reliable for personalized recommendations based on changing data.
2. Retrieval-Augmented Generation (RAG)

RAG retrieves the latest information from external data sources before generating a response. In this use case, it can retrieve current restaurant menus, item availability, prices, and dietary information.

Advantages:

Provides up-to-date menu information.
No need to retrain the model whenever restaurant data changes.
Supports personalized recommendations by retrieving user order history and dietary preferences from a database.
Scales well for a large and diverse user base.

Limitations:

Requires maintaining a retrieval system and knowledge base.
Slightly higher response latency than prompt engineering.
3. Fine-Tuning

Fine-tuning trains the model on domain-specific examples so that it learns the preferred behavior of the application.

Advantages:

Produces consistent responses in the food delivery domain.
Can improve recommendation quality and customer support interactions.
Useful for specialized tasks such as complaint handling or intent classification.

Limitations:

Expensive and time-consuming to train.
Not suitable for frequently changing menu data because retraining would be required whenever menus change.
Less flexible for dynamic information.
Recommended Approach

The best solution is a combination of RAG and Prompt Engineering.

RAG should retrieve the latest menu availability, dietary information, restaurant details, and user order history.
Prompt Engineering should instruct the language model to generate personalized recommendations while considering dietary restrictions and the retrieved information.

Fine-tuning can be added later only if the platform requires highly specialized behavior, such as advanced recommendation ranking or customer support automation, but it should not be used to store dynamic menu data.

Why the Other Approaches Fall Short
Prompt Engineering alone cannot access real-time menu updates or maintain personalized knowledge without supplying all information in every prompt.
Fine-Tuning alone is unsuitable because menu availability changes daily, making frequent retraining impractical and expensive.
RAG combined with Prompt Engineering provides the latest information, supports personalization, scales to a diverse user base, and avoids unnecessary retraining.

# SCENARIO :-2 


You are working on an AI team at a food delivery company and have been asked to
fine-tune a large language model to understand food-specific customer service
queries. Your infrastructure has a single consumer-grade GPU with 8 GB VRAM, and full
fine-tuning of even a 7B-parameter model is not feasible.


Question: Explain what LoRA (Low-Rank Adaptation) modifies in a pre-trained model compared
to full fine-tuning. Why does this make it viable under severe VRAM constraints? What trade-off, if
any, does the team accept by using LoRA instead of full fine-tuning?

What LoRA Modifies

LoRA (Low-Rank Adaptation) is a parameter-efficient fine-tuning technique. Instead of updating all the weights of a pre-trained language model, LoRA freezes the original model parameters and inserts small trainable low-rank matrices into selected layers (typically the attention layers). During training, only these newly added matrices are updated, while the original model weights remain unchanged.

In contrast, full fine-tuning updates every parameter in the model, which requires significantly more GPU memory and computational resources.

Why LoRA is Suitable for an 8 GB VRAM GPU

LoRA is practical under limited VRAM because:

Only a small fraction of the model parameters are trainable.
The original model weights remain frozen, reducing memory usage.
Fewer gradients and optimizer states need to be stored during training.
Training is faster and requires less computational power.
Multiple LoRA adapters can be created for different tasks without storing multiple copies of the full model.

Because of these advantages, a 7B-parameter model can often be adapted on a consumer GPU with around 8 GB VRAM, whereas full fine-tuning would typically require much more memory.

Trade-Offs of Using LoRA

The team accepts a few trade-offs when choosing LoRA instead of full fine-tuning:

LoRA may achieve slightly lower performance than full fine-tuning for some highly specialized tasks.
Only selected layers are adapted, so the model has less flexibility to learn large behavioral changes.
If the task requires major modifications to the model's knowledge or reasoning, full fine-tuning may provide better results.
During inference, the LoRA adapter must be loaded along with the base model (unless the adapter is merged into the model beforehand).

# SCENARIO :-3 


You are debugging a fine-tuned complaint classifier deployed by a food delivery
company. The model was trained on 6,000 labelled complaint tickets and performs well
on common issues — wrong item delivered, cold food — but fails on rare complaints like
'delivery partner was rude during a religious event' that were absent from training data.



Question: Identify two fundamental limitations of fine-tuning that this scenario exposes. For
each limitation, explain whether re-fine-tuning with more data, switching to RAG, or using
few-shot prompting in the system prompt would be the more appropriate remedy — and state
the condition under which your chosen remedy is valid.

The scenario highlights two important limitations of fine-tuning.

1. Limited Generalization to Unseen Cases

The model was trained on common complaint categories but has never seen rare complaints such as "delivery partner was rude during a religious event." As a result, it struggles to correctly classify these unseen situations.

Appropriate Remedy: Re-fine-tuning with more labelled data

If examples of these rare complaint types are collected and labelled, the model can be fine-tuned again to learn these new patterns and improve its performance.

Condition:
This approach is effective only if a sufficient amount of high-quality labelled training data for the new complaint categories is available.

2. Static Knowledge After Training

A fine-tuned model only learns from the data available during training. It cannot automatically adapt to new complaint categories, updated company policies, or changing business rules without another training cycle.

Appropriate Remedy: RAG (Retrieval-Augmented Generation)

RAG allows the model to retrieve the latest complaint handling guidelines, company policies, or support documentation at inference time. This enables the system to use current information without retraining the model whenever policies change.

Condition:
RAG is appropriate when the required knowledge exists in external documents or databases that can be retrieved during inference.

When Few-Shot Prompting is Appropriate

Few-shot prompting can also help if only a small number of new complaint examples are available and retraining is not practical. By including representative examples in the system prompt, the model can better understand similar complaint patterns.

Condition:
Few-shot prompting is effective when the task is relatively simple, the number of new examples is small, and the base language model already has sufficient general reasoning ability.

Food Item: Masala Dosa

Verify that delivery is possible at the requested time.
Estimate the delivery time.
If all conditions are satisfied, proceed with placing the order.

The agent resolves these ambiguities by using stored user information or asking the user for clarification if needed.

check_restaurant_availability("Saravana Bhavan", "7 PM")

# SCENARIO:-5 



You are building a multi-agent system for a food delivery operations team. Agent 1
tracks live delivery statuses, Agent 2 handles customer complaints, and Agent 3
generates daily revenue and delay reports. A customer complaint arrives about a
45-minute delay on Order #FD7823, which is still in transit and will likely affect that day's
delay statistics.


Question: How should task delegation be structured so that Agent 2 (Complaint Handler) can
resolve the customer's immediate concern while simultaneously informing Agent 1 (Tracker) and
Agent 3 (Reporter) about the impacted order? Describe the communication pattern between
agents and identify at least two failure modes that a poorly coordinated multi-agent system
would exhibit in this exact scenario.

In this scenario, the system should use a coordinator-based multi-agent architecture, where each agent has a specific responsibility and communicates through a central coordinator or shared event system.

Agent Responsibilities
Agent 1 – Delivery Tracker
Monitors the live status of Order FD7823.
Updates delivery location, ETA, and delay information.
Agent 2 – Complaint Handler
Receives the customer's complaint.
Responds to the customer and creates a complaint record.
Notifies other agents about the affected order.
Agent 3 – Reporter
Collects delivery and complaint data.
Updates daily delay and revenue reports.
Communication Pattern
Step 1: Complaint Received

The customer reports a 45-minute delay for Order FD7823.

Agent 2 receives the complaint and records it.

Step 2: Notify Agent 1

Agent 2 sends the order ID (FD7823) to Agent 1.

Agent 1:

Verifies the live delivery status.
Calculates the updated ETA.
Returns the latest tracking information.
Step 3: Respond to Customer

Using the updated information from Agent 1, Agent 2:

Explains the reason for the delay (if available).
Provides the revised ETA.
Offers an apology or compensation according to company policy.
Step 4: Notify Agent 3

After resolving the customer's immediate concern, Agent 2 also informs Agent 3 that Order FD7823 experienced a delay and generated a customer complaint.

Agent 3:

Records the delayed order.
Includes it in the daily delay statistics.
Updates operational reports and performance metrics.
Communication Flow
Customer
    │
    ▼
Agent 2 (Complaint Handler)
    │
    ├────────► Agent 1 (Delivery Tracker)
    │              │
    │              ▼
    │      Updated ETA & Status
    │
    ├────────► Customer
    │      (Resolution & Updated ETA)
    │
    └────────► Agent 3 (Reporter)
                   │

         Update Delay & Revenue Reports
Failure Modes in a Poorly Coordinated System
1. Inconsistent Information

If Agent 2 does not communicate with Agent 1, the customer may receive an outdated or incorrect delivery status or ETA.

Impact:

Poor customer experience.
Loss of trust.
Repeated complaints.
2. Incorrect Reporting

If Agent 3 is not informed about the delayed order, the daily operational reports will not include the delay.

Impact:

Inaccurate delay statistics.
Incorrect performance analysis.
Poor management decisions based on incomplete data.
Additional Failure Mode
Duplicate or Conflicting Actions

If multiple agents independently handle the same complaint without coordination:

The customer may receive conflicting responses.
Multiple refunds or compensation requests could be generated.
Duplicate complaint records may be created.

# SCENARIO:- 6


You are optimising the customer support agent for a food delivery platform using the
ReAct (Reason + Action) pattern. A user sends the query: 'My order #FD4521 was 40
minutes late and the food arrived cold. I want either a full refund or a significant
discount on my next order.'


Question: Walk through at least two complete Reason › Action › Observation cycles that this
ReAct agent would execute to handle this query. For each cycle, state what the agent reasons,
what action it takes (tool called), and what it observes from the result. Explain why separating
the reasoning step from the action step is critical when the agent is making customer-facing
decisions that have financial consequences.

The ReAct (Reason → Action → Observation) pattern allows the agent to reason about the problem, use tools to gather facts, and then make an informed decision instead of guessing.

Cycle 1

Reason:
The customer claims that Order FD4521 was 40 minutes late and the food arrived cold. Before deciding on compensation, the agent must verify the order details and delivery status.

Action (Tool Called):
lookup_order_status("FD4521")

Observation:

Order ID: FD4521
Status: Delivered
Delivery Delay: 40 minutes
Delivery completed successfully

The complaint about the delivery delay is confirmed.

Cycle 2

Reason:
Since the delay has been verified, the agent must determine whether the order qualifies for compensation according to the company's refund policy.

Action (Tool Called):
check_refund_eligibility("FD4521")

Observation:

Delay exceeds the refund threshold.
Customer is eligible for a full refund.
A full refund takes priority over a discount coupon.

The agent now has enough verified information to respond confidently.

Final Response to the Customer

"Thank you for your patience. We verified that Order FD4521 was delivered 40 minutes late, making it eligible for compensation under our policy. We have approved a full refund for your order, and the amount will be processed shortly. We apologize for the inconvenience caused."

Why Separating Reasoning from Action is Important

Separating the Reason step from the Action step is essential because customer-facing decisions can have financial consequences.

The Reason step determines what information is needed before making a decision.
The Action step retrieves verified information using trusted tools or databases.
The Observation step confirms the results, allowing the agent to base its response on facts rather than assumptions.

Without this separation, the agent might:

Approve refunds without verifying eligibility.
Deny compensation even when the customer qualifies.
Provide incorrect delivery information.
Cause financial losses or inconsistent customer service.

By following the Reason → Action → Observation cycle, the agent makes accurate, transparent, and policy-compliant decisions, ensuring fairness for both the customer and the food delivery company.
