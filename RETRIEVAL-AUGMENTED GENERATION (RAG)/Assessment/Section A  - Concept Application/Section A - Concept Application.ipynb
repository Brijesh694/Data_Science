### SCENARIO 1 :-  You are building a customer support chatbot for a food delivery app. During testing, you notice that the chatbot gives inconsistent answers — sometimes it refuses valid refund requests, and sometimes it approves them without verifying order status. Your team traces the problem to vague prompts that do not specify the expected output format or the decision rules the LLM should follow.

Question: Explain what distinguishes a clear prompt from an unclear one in this context.
Describe at least two specific changes you would make to the existing prompts to produce
consistent, reliable responses from the LLM, and justify why each change matters.

A clear prompt gives the LLM specific instructions about its role, decision-making process, required information, and output format. An unclear prompt is vague, lacks rules, and leaves the model to guess how to respond. This can lead to inconsistent decisions, such as approving invalid refunds or rejecting valid ones.

Example of an Unclear Prompt

"Help customers with refund requests."

This prompt does not explain:

When refunds should be approved or denied.
What information must be checked.
How the response should be formatted.
Example of a Clear Prompt

"You are a customer support assistant for a food delivery app. Before approving a refund, verify the order status and refund policy. Approve refunds only if the order was cancelled by the restaurant or the food was not delivered. If information is missing, ask the customer for the order ID. Respond using the format: Decision, Reason, and Next Steps."

Two Specific Changes to Improve the Prompt
1. Add Explicit Decision Rules

Change:
Include clear refund eligibility conditions in the prompt.

Example:

Approve refund if the order was not delivered.
Approve refund if the restaurant cancelled the order.
Reject refund if the order was delivered successfully.
Request more information if the order status is unknown.

Why it matters:
The LLM follows predefined business rules instead of making assumptions. This reduces inconsistent decisions and ensures all customers are treated fairly.

2. Define a Structured Output Format

Change:
Require every response to follow a fixed structure.

Example Format:

Decision: Approved / Rejected / Need More Information
Reason: <Explanation>
Next Steps: <Action for customer>

Why it matters:
A structured format ensures consistent responses, makes answers easier to review, and prevents the model from skipping important information.

Additional Improvement

Require Verification Before Decision

Example:

"Do not approve or reject a refund until the order status and order ID have been verified."

Why it matters:
This prevents the chatbot from approving refunds without checking necessary details, improving reliability and reducing errors.

### SCENARIO:-2 You are designing a complaint classifier for a food delivery platform. The classifier must label incoming customer messages into one of four categories: 'Late Delivery', 'Wrong Item', 'Missing Item', and 'Poor Quality'. You have 200 labelled examples available, but the LLM you are using has a limited context window and cannot accept all of them at once

Zero-Shot Prompting vs Few-Shot Prompting

Zero-Shot Prompting

The LLM receives only the task description and category definitions.
No training examples are provided.

Example prompt:

Classify the customer complaint into one of these categories: Late Delivery, Wrong Item, Missing Item, Poor Quality.

Advantages:

Uses very little context space.
Fast and simple to implement.
Suitable when categories are clear and distinct.

Disadvantages:

May produce inconsistent classifications.
Can struggle with ambiguous customer messages.
Lower accuracy compared to few-shot prompting.

Few-Shot Prompting

The LLM receives the task description along with several labelled examples.
Examples demonstrate the correct mapping between complaints and categories.

Example:

Message: "My order arrived 45 minutes late."
Category: Late Delivery

Message: "I ordered pizza but received a burger."
Category: Wrong Item

Message: "The fries were missing from my order."
Category: Missing Item

Advantages:

Improves classification accuracy.
Helps the model understand category boundaries.
Produces more consistent outputs.

Disadvantages:

Consumes context window space.
Too many examples may exceed the model's input limit.
Which Approach Would I Choose?

Given the constraints:

There are 200 labelled examples available.
The LLM has a limited context window and cannot accept all examples.

I would choose few-shot prompting.

Reason:

Complaint classification is a structured task with only four categories.
A small set of carefully selected examples (e.g., 2–5 examples per category) can significantly improve accuracy over zero-shot prompting.
It provides guidance without exceeding the context limit.
Using all 200 examples is unnecessary because many examples may be repetitive.
Factors Determining How Many Examples to Include
Context Window Size
Larger context windows allow more examples.
Limited context requires selecting only the most representative examples.
Task Complexity
Simple categories need fewer examples.
Ambiguous or overlapping categories require more examples.
Category Coverage
Each category should be represented.
Include examples for all four complaint types.
Example Diversity
Use varied wording and complaint styles.
Diverse examples improve generalization.
Prompt Length vs Performance Trade-off
More examples may improve accuracy initially.
Beyond a certain point, additional examples provide little benefit while consuming valuable context.

### SCENARIO:-3  You are developing an AI assistant that helps delivery operations staff calculate the optimal stop sequence for a delivery agent given five order addresses, estimated traffic delays, and priority levels. When you send a direct question to the LLM asking for the best route, the answers are frequently incorrect and contain no explanation of the reasoning

Question: Explain how Chain-of-Thought (CoT) prompting would improve the quality of the
assistant's output for this task. Write a brief example showing how you would restructure the
prompt to invoke step-by-step reasoning, and identify one limitation of this technique.

Chain-of-Thought (CoT) Prompting

Chain-of-Thought (CoT) prompting encourages the LLM to solve a problem by breaking it into smaller reasoning steps before giving the final answer. Instead of directly asking for the best delivery route, the prompt asks the model to analyze traffic delays, priority levels, and distances step by step.

How CoT Improves Output Quality
Structured Reasoning – The model evaluates each factor systematically rather than guessing.
Better Accuracy – Breaking the task into steps reduces reasoning errors.
Transparency – Operations staff can understand why a route was chosen.
Consistency – Similar delivery scenarios are handled more reliably.
Example of a CoT Prompt

Direct Prompt (Less Effective):

Find the best delivery route for these five orders and return the route.

Chain-of-Thought Prompt (Improved):

You are a delivery route planner.

Given five delivery addresses, traffic delays, and priority levels:

1. List all delivery stops.
2. Identify high-priority orders.
3. Estimate travel time between stops considering traffic.
4. Compare possible delivery sequences.
5. Select the route that minimizes total delivery time while serving high-priority orders first.
6. Explain your reasoning step by step.
7. Provide the final recommended route.

Data:
Order A: Priority High, Traffic Delay 5 min
Order B: Priority Low, Traffic Delay 15 min
Order C: Priority High, Traffic Delay 10 min
Order D: Priority Medium, Traffic Delay 8 min
Order E: Priority Low, Traffic Delay 12 min
Limitation of Chain-of-Thought Prompting

A major limitation is that CoT does not guarantee correct reasoning. The model may generate logical-looking steps that still contain mistakes, especially for complex optimization problems such as route planning. It also increases response length and consumes more tokens, which can increase cost and latency.

Why RAG is Better Than Fine-Tuning for This Use Case

For a 150-page delivery policy handbook that is updated every quarter, Retrieval-Augmented Generation (RAG) is a better choice than fine-tuning.

1. Handles Frequent Updates Easily
In a RAG system, the handbook is stored in a knowledge base or vector database.
When policies change, you simply update the documents and re-index them.
No need to retrain the model.

Fine-Tuning:

Every major handbook update would require collecting new training data and re-training or re-fine-tuning the model.
This increases time, cost, and maintenance effort.
2. Provides Up-to-Date and Accurate Answers
RAG retrieves the most relevant sections of the latest handbook before generating an answer.
Responses are grounded in current policy documents, reducing outdated information and hallucinations.

Fine-Tuning:

The model only knows what was present during training.
If payout rules or SLA terms change after training, the model may provide outdated answers.
3. Lower Cost and Faster Deployment
RAG only requires document chunking, embedding, and retrieval.
Updating knowledge is much cheaper than repeated fine-tuning.

Fine-Tuning:

Requires training resources, compute power, and validation after every update.
4. Better Transparency
RAG can show the exact handbook sections used to answer a question.
Employees can verify the source of the information.

Fine-Tuning:

Answers come from model parameters, making it difficult to identify the exact source.

| Feature                   | RAG       | Fine-Tuning             |
| ------------------------- | --------- | ----------------------- |
| Frequent document updates | Easy      | Difficult               |
| Up-to-date information    | Yes       | No, requires retraining |
| Cost of maintenance       | Low       | Higher                  |
| Source citation           | Available | Limited                 |
| Deployment speed          | Faster    | Slower                  |


When Fine-Tuning Would Be Better

Fine-tuning would be the better choice if the goal was to teach the model a specialized behavior or style rather than provide knowledge from frequently changing documents.

Example:
A customer-support assistant that must always respond in a specific company tone, follow a fixed conversation structure, and perform classification tasks consistently. Since the behavior changes rarely, fine-tuning can improve performance and consistency without requiring document retrieval.

### SCENARIO :-5 You are building a RAG pipeline to let customers search across 500 restaurant menus stored as PDF files. Each menu averages 8 pages. During evaluation you find that retrieved chunks often cut mid-sentence or omit critical context — for example, a price appears in a sentence one paragraph after the dish name, but both rarely appear in the same chunk.


Question: Explain the role of chunk size and chunk overlap in a RAG pipeline. Describe how you
would adjust these two parameters to fix the retrieval quality problems described, and identify
one trade-off your adjustment introduces.

Role of Chunk Size and Chunk Overlap in a RAG Pipeline

In a RAG (Retrieval-Augmented Generation) system, documents are divided into smaller pieces called chunks before being converted into embeddings and stored in a vector database.

1. Chunk Size
Chunk size determines how much text is included in each chunk.
If chunks are too small, important information may be split across multiple chunks.
If chunks are too large, retrieval becomes less precise because each chunk contains too much information.
2. Chunk Overlap
Chunk overlap means that consecutive chunks share some text.
This helps preserve context when information spans chunk boundaries.
Overlap reduces the chance of important details being separated into different chunks.
Problem in the Scenario

The retrieved chunks:

Cut off in the middle of sentences.
Separate dish names from their prices.
Miss important contextual information spread across paragraphs.

Example:

Chunk 1:
Paneer Tikka served with mint chutney...

Chunk 2:
Special Offer Price: ₹220

The dish name and price are stored in different chunks, making retrieval less effective.

Recommended Adjustments
Increase Chunk Size

Use larger chunks so related information remains together.

Example:

Current chunk size: 100 words
New chunk size: 300–500 words

This increases the likelihood that:

Dish name
Description
Price
Special offers

all remain within the same chunk.

Increase Chunk Overlap

Add overlap between neighboring chunks.

Example:

Current overlap: 0–20 words
New overlap: 50–100 words

This ensures that information near chunk boundaries appears in multiple chunks.

Trade-Off

The main trade-off is higher storage and computational cost.

Larger chunks increase embedding size.
More overlap creates duplicate text across chunks.
Vector databases require more storage.
Retrieval and indexing become slower.

For example:

Chunk size = 500 words
Overlap = 100 words

may significantly improve context retention, but it increases the number of tokens stored and processed.
